### Gộp các file của mỗi tỉnh thành 1 file thống nhất 

```python
# Đường dẫn đến thư mục chứa file
folder_path = "dataset"
all_files = os.listdir(folder_path)

# Lọc các file csv (nếu có file khác loại)
csv_files = [f for f in all_files if f.endswith(".csv")]

# Danh sách chứa các dataframe
df_list = []

for file in csv_files:
    file_path = os.path.join(folder_path, file)
    
    # Đọc file
    df = pd.read_csv(file_path)
    
    # Thêm mã tỉnh từ tên file (ví dụ: 79.csv → mã tỉnh = 79)
    province_code = os.path.splitext(file)[0]
    df["province_code"] = province_code
    
    df_list.append(df)

# Gộp tất cả lại
combined_df = pd.concat(df_list, ignore_index=True)

# Xuất ra file tổng
combined_df.to_csv("weather_combined.csv", index=False)

print("Đã gộp xong")
```
